```python
import os
import sys

sys.path.append(os.path.abspath(os.path.join("..")))
from IPython.display import Image, display
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.runnables import RunnableConfig
from langgraph.checkpoint.memory import MemorySaver
from langgraph.graph import START, MessagesState, StateGraph
from langgraph.prebuilt import ToolNode, tools_condition

from utils.model import Models
```

```python
model = Models()
llm = model.create_groq_model()
```

```python
def multiply(a: int, b: int) -> int:
    """Multiply the two arguments and return the result
    Args:
    a:First Argument
    b:Second Argument
    """
    return a * b


def addition(a: int, b: int) -> int:
    """Add the two arguments and return the result
    Args:
    a:First Argument
    b:Second Argument
    """
    return a + b


def subtract(a: int, b: int) -> int:
    """Subtract the two arguments and return the result
    Args:
    a:First Argument
    b:Second Argument
    """
    return a - b


tools = [addition, subtract, multiply]
```

```python
llm_with_tools = llm.bind_tools(tools)
```

```python
def assistant(state: MessagesState):
    sys_msg = SystemMessage(
        content="You are a helpful assistant tasked to do only arithmetic operations"
    )
    return {"messages": llm_with_tools.invoke([sys_msg] + state["messages"])}
```

```python
builder = StateGraph(MessagesState)
builder.add_node("assistant", assistant)
builder.add_node("tools", ToolNode(tools))
builder.add_edge(START, "assistant")
builder.add_conditional_edges("assistant", tools_condition)
builder.add_edge("tools", "assistant")
graph = builder.compile(interrupt_before=["assistant"], checkpointer=MemorySaver())
display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
input_thread_1: RunnableConfig = {"configurable": {"thread_id": "1"}}
```

```python
input_msg = HumanMessage(content="Mutiply 3 and 6")
```

```python
for event in graph.stream(
    {"messages": [input_msg]}, config=input_thread_1, stream_mode="values"
):
    event["messages"][-1].pretty_print()
```

```python
graph.update_state(
    input_thread_1, {"messages": [HumanMessage(content="No multiply 6 and 7")]}
)
```

```python
new_state = graph.get_state(input_thread_1)
for m in new_state.values["messages"]:
    m.pretty_print()
```

```python
for event in graph.stream(None, input_thread_1, stream_mode="values"):
    event["messages"][-1].pretty_print()
```

### _Awaiting user Input_
- In the graph `human_feedback` allows user to add feedback directly to the state 
- Add breakpoint `interrupt_before` right before the `human_feedback` node 
- We can add a dummy node, which only present in the graph so we can get user input 

```python
def human_feedback(state: MessagesState):
    pass
```

```python
def assistant(state: MessagesState):
    sys_msg = [
        SystemMessage(
            content="You are a helpful assistant tasked to do only arithmetic operations"
        )
    ]
    return {"messages": llm_with_tools.invoke(sys_msg + state["messages"])}
```

```python
builder_1 = StateGraph(MessagesState)
builder_1.add_node("assistant", assistant)
builder_1.add_node("tools", ToolNode(tools))
builder_1.add_node("human_feedback", human_feedback)
builder_1.add_edge(START, "human_feedback")
builder_1.add_edge("human_feedback", "assistant")
builder_1.add_conditional_edges("assistant", tools_condition)
builder_1.add_edge("tools", "human_feedback")
graph1 = builder_1.compile(
    interrupt_before=["human_feedback"], checkpointer=MemorySaver()
)
display(Image(graph1.get_graph().draw_mermaid_png()))
```

```python
for m in graph1.stream({"messages": input_msg}, input_thread_1, stream_mode="values"):
    m["messages"][-1].pretty_print()

user_input = input("How do you wann to proceed?")
graph1.update_state(input_thread_1, {"messages": user_input}, as_node="human_feedback")

for m in graph1.stream(None, input_thread_1, stream_mode="values"):
    m["messages"][-1].pretty_print()
```
