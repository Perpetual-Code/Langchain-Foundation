```python
from operator import add
from typing import Annotated, TypedDict
```

```python
class Logs(TypedDict):
    _id: str
    questions: str
    answer: str
    grader: str | None
    grade: int | None
    feedback: str | None
```

```python
class FailureAnalysisState(TypedDict):
    cleaned_logs: list[Logs]
    fa_summary: str
    failures: list[Logs]
    processed_logs: list[str]


class FailureAnalysisOutputState(TypedDict):
    fa_summary: str
    processed_logs: list[str]
```

```python
class QuestionSummarizationState(TypedDict):
    qa_summary: list[str]
    processed_logs: list[Logs]
    report: str
    cleaned_logs: list[Logs]


class QuestionSummaryOutputState(TypedDict):
    processed_logs: list[Logs]
    report: str
```

```python
class Lead(TypedDict):
    _id: str
    lead_name: str
    email: str
    social_media_links: str
    icp_score: int | None  # None because initially it will be 0


class CompanyResearchAnalysisState(TypedDict):
    company_name: str
    company_social_media: str
    ca_report: str
    ca_score: int


class CompanyAnalysisOutputState(TypedDict):
    ca_report: str
    ca_score: int


class PersonalAnalysisState(TypedDict):
    name: str
    pa_social_media: str
    pa_report: str
    pa_score: int


class PersonalAnalysisOutputState(TypedDict):
    pa_report: str
    pa_score: int


class EntryGraphState(TypedDict):
    lead: list[Lead]


class ExitgraphState(TypedDict):
    # INFO sum of ca_score,pa_score and other factors(I  am thinking but as I work through this,it will be more clear)
    icp_score: Annotated[int, add]
    reports: Annotated[list, add]
```
